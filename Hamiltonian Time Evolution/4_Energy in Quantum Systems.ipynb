```python
import pennylane as qp
qml = qp
import numpy as np
```

```python
n_bits = 2
dev = qp.device("default.qubit", wires=n_bits)

@qp.qnode(dev)
def zz_circuit(alpha, time, init):
    """Circuit for evolving two electrons with a ZZ interaction.
    
    Args:
        alpha (float): The strength of the interaction.
        time (float): The time we evolve the electron wavefunction for.
        init (numpy.array(int)): An initial state specified by two bits [x, y]. Prepare the
            system in this state prior to applying the time evolution circuit.

    Returns: 
        array[float]: Probabilities for observing different outcomes.
    """

    ##################
    # YOUR CODE HERE #
    ##################
    qp.BasisState(init, wires=[0,1])
    qp.CNOT(wires=[0,1])
    qp.RZ(2*alpha*time, wires=1)
    qp.CNOT(wires=[0,1])
    
    return qp.probs(wires=range(n_bits))
```

```python
n_bits = 2
dev = qp.device("default.qubit", wires=n_bits)

@qp.qnode(dev)
def ising_circuit(alpha, time, init):
    """Circuit for evolving two electrons with a ZZ interaction
    using an Ising gate
    
    Args:
        alpha (float): The strength of the interaction.
        time (float): The time we evolve the electron wavefunction for.
        init (numpy.array(int)): An initial state specified by two bits [x, y]. Prepare the
            system in this state prior to applying the time evolution circuit.

    Returns: 
        np.tensor: Output state.
    """
    ##################
    # YOUR CODE HERE #
    ##################
    qp.BasisState(init, wires=range(2))
    qp.IsingZZ(2*alpha*time, wires=[0,1])
    
    return qp.state()
```

```python
n_bits = 2
dev = qp.device("default.qubit", wires=n_bits)

@qp.qnode(dev)
def ZZ_evolve(alpha, time, init):
    """Circuit for evolving two electrons with a ZZ interaction
    using qp.evolve
    
    Args:
        alpha (float): The strength of the interaction.
        time (float): The time we evolve the electron wave function for.
        init (numpy.array(int)): An initial state specified by two bits [x, y]. Prepare the
            system in this state prior to applying the time evolution circuit.

    Returns: 
        np.tensor: Output state.
    """
    ##################
    # YOUR CODE HERE #
    ##################
    qp.BasisState(init, wires=range(2))
    qp.evolve(alpha * qp.PauliZ(0) @ qp.PauliZ(1))
    
    return qp.state()
```

```python
n_bits = 5
dev = qp.device("default.qubit", wires=n_bits)
    
##################
# YOUR CODE HERE #
##################
coeffs = [1, 1, 1, 1] # MODIFY THIS
obs = [qp.PauliZ(0)@qp.PauliZ(1), qp.PauliZ(1)@qp.PauliZ(2),
       qp.PauliZ(1)@qp.PauliZ(3), qp.PauliZ(3)@qp.PauliZ(4)] # MODIFY THIS
H = qp.dot(coeffs, obs)

@qp.qnode(dev)
def energy(init):
    """Circuit for measuring expectation value of Hamiltonian in a given state.
    
    Args:
        init (numpy.array(int)): An initial computational basis state, specified by five bits.

    Returns: 
        float: Expectation value of the Hamiltonian H.
    """
    qp.BasisState(init, wires=range(n_bits))
    
    return qp.expval(H)
```
