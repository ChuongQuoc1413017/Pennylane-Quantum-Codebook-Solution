```python
import pennylane as qp
import numpy as np
```

```python
dev = qp.device("default.qubit", wires=2)

# Starting from the state |00>, implement a PennyLane circuit
# to construct each of the Bell basis states.


@qp.qnode(dev)
def prepare_psi_plus():
    ##################
    # YOUR CODE HERE #
    ##################

    # PREPARE (1/sqrt(2)) (|00> + |11>)
    qp.Hadamard(wires = 0)
    qp.CNOT(wires = [0, 1])

    return qp.state()


@qp.qnode(dev)
def prepare_psi_minus():
    ##################
    # YOUR CODE HERE #
    ##################

    # PREPARE (1/sqrt(2)) (|00> - |11>)
    qp.PauliX(wires = 0)
    qp.Hadamard(wires = 0)
    qp.CNOT(wires = [0, 1])

    return qp.state()


@qp.qnode(dev)
def prepare_phi_plus():
    ##################
    # YOUR CODE HERE #
    ##################

    # PREPARE  (1/sqrt(2)) (|01> + |10>)
    qp.Hadamard(wires = 1)
    qp.PauliX(wires = 0)
    qp.CNOT(wires = [1, 0])

    return qp.state()


@qp.qnode(dev)
def prepare_phi_minus():
    ##################
    # YOUR CODE HERE #
    ##################

    # PREPARE  (1/sqrt(2)) (|01> - |10>)
    qp.PauliX(wires = 1)
    qp.Hadamard(wires = 1)
    qp.PauliX(wires = 0)
    qp.PauliX(wires = 1)
    qp.CNOT(wires = [1, 0])

    return qp.state()


psi_plus = prepare_psi_plus()
psi_minus = prepare_psi_minus()
phi_plus = prepare_phi_plus()
phi_minus = prepare_phi_minus()

# Uncomment to print results
print(f"|ψ_+> = {psi_plus}")
print(f"|ψ_-> = {psi_minus}")
print(f"|ϕ_+> = {phi_plus}")
print(f"|ϕ_-> = {phi_minus}")
```
