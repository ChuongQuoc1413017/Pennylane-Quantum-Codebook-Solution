```python
import pennylane as qp
qml = qp
import numpy as np
```

```python
num_wires = 1
dev = qp.device("default.qubit", wires=num_wires)

@qp.qnode(dev)
def one_qubit_QFT(basis_id):
    """A circuit that computes the QFT on a single qubit. 
    
    Args:
        basis_id (int): An integer value identifying 
            the basis state to construct.
    
    Returns:
        array[complex]: The state of the qubit after applying QFT.
    """
    # Prepare the basis state |basis_id>
    bits = [int(x) for x in np.binary_repr(basis_id, width=num_wires)]
    qp.BasisState(bits, wires=[0])

    ##################
    # YOUR CODE HERE #
    ##################
    qp.Hadamard(wires = 0)
    
    return qp.state()
```

```python
num_wires = 2
dev = qp.device("default.qubit", wires=num_wires)

@qp.qnode(dev)
def two_qubit_QFT(basis_id):
    """A circuit that computes the QFT on two qubits using qp.QubitUnitary. 
    
    Args:
        basis_id (int): An integer value identifying the basis state to construct.
    
    Returns:
        array[complex]: The state of the qubits after the QFT operation.
    """
    
    # Prepare the basis state |basis_id>
    bits = [int(x) for x in np.binary_repr(basis_id, width=num_wires)]
    qp.BasisState(bits, wires=[0, 1])
    
    ##################
    # YOUR CODE HERE #
    ##################
    #Define a function that raises omega to the nth power
    U = (1/2)*np.array([[1, 1, 1, 1], 
                        [1, complex(0,1), -1, complex(0,-1)], 
                        [1, -1, 1, -1], 
                        [1, complex(0,-1), -1, complex(0,1)]])
    qp.QubitUnitary(U, wires=[0, 1])

    return qp.state()
```

```python
num_wires = 2
dev = qp.device("default.qubit", wires=num_wires)

@qp.qnode(dev)
def decompose_two_qubit_QFT(basis_id):
    """A circuit that computes the QFT on two qubits using elementary gates.
    
    Args:
        basis_id (int): An integer value identifying the basis state to construct.
    
    Returns:
        array[complex]: The state of the qubits after the QFT operation.
    """
    # Prepare the basis state |basis_id>
    bits = [int(x) for x in np.binary_repr(basis_id, width=num_wires)]
    qp.BasisState(bits, wires=[0, 1])
    
    ##################
    # YOUR CODE HERE #
    ##################
    qp.Hadamard(wires=0)
    qp.ctrl(qp.S, control=1, control_values=1)(wires=0)
    qp.Hadamard(wires=1)
    qp.SWAP(wires=[0,1])

    return qp.state()
```
